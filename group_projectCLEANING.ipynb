## Importing Libraries, Data and Warnings


```python
#importing libraries
#in the terminal write it as pip install (library to install)
import pandas as pd #write this so code recognises pd

#import warnings
import warnings
warnings.simplefilter("ignore", category=FutureWarning)
warnings.simplefilter("ignore", category=DeprecationWarning)


#import data
test=pd.read_csv("Rossman_test.csv")
store=pd.read_csv("Rossman_store.csv")
train=pd.read_csv("Rossman_train.csv") #there is a mix type in column 7
```

## Store Data

```python
store.shape
```

```python
store.head()
```

```python
store.describe()
```

```python
#fill in missing values

#filled missing values with mean
store["CompetitionDistance"]=store["CompetitionDistance"].fillna(store["CompetitionDistance"].mean())

#fill in any missing years and months with 0
store["CompetitionOpenSinceMonth"]=store["CompetitionOpenSinceMonth"].fillna(0).astype(int)
store["CompetitionOpenSinceYear"]=store["CompetitionOpenSinceYear"].fillna(0).astype(int)

#the store does not have values for promo 2 and promo interval so fill in missing values with 0
store["Promo2SinceWeek"]=store["Promo2SinceWeek"].fillna(0).astype(int)
store["Promo2SinceYear"]=store["Promo2SinceYear"].fillna(0).astype(int)
store["PromoInterval"]=store["PromoInterval"].fillna("None")
```

```python
store.to_csv("store_cleaned.csv", index=False)
```

## Test Data

```python
test.shape
```

```python
test.head
```

```python
test.describe()
```

```python
#add the date for each month
test["Date"]=pd.to_datetime(test["Date"])

#Open has some missing values so replace with 1 for open as most days that are not filled are weekdays
test["Open"]=test["Open"].fillna(1).astype(int)
```

```python
test.to_csv("test_cleaned.csv", index=False)
```
